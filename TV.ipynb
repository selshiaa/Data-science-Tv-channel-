Data Cleaning

26.	Standardize the Channel_Name values.

```python
# Remove extra spaces
df['Channel_Name'] = df['Channel_Name'].str.strip()

# Standardize channel names
df['Channel_Name'] = df['Channel_Name'].replace({
    'SunTV': 'Sun TV',
    'ZeeTamil': 'Zee Tamil',
    'colors tamil': 'Colors Tamil',
    'COLORS TAMIL': 'Colors Tamil',
    'ColorsTamil': 'Colors Tamil',
    'vijay tv': 'Vijay TV'
})

# Check the standardized values
print(df['Channel_Name'].unique())
```

27.	Handle missing values appropriately.

```python
df.isnull().sum()
```

```python
numeric_cols = df.select_dtypes(include='number').columns

for col in numeric_cols:
    df[col] = df[col].fillna(df[col].median())  #For numeric columns, use the median
```

```python
categorical_cols = df.select_dtypes(include='object').columns

for col in categorical_cols:
    df[col] = df[col].fillna(df[col].mode()[0])  #For categorical columns, use the mode:
```

```python
print(df.isnull().sum())
```

29.	Correct invalid Avg_Rating values.

```python
import pandas as pd

# Example dataset
data = {
    "Channel_ID": [1, 2, 3, 4],
    "Channel_Name": ["A", "B", "C", "D"],
    "Avg_Rating": [4.5, -1, 6.2, None]  # invalid values included
}

df = pd.DataFrame(data)

# ✅ Step 1: Define valid range (0–5)
valid_min, valid_max = 0, 5

# ✅ Step 2: Identify invalid values
invalid = df[(df["Avg_Rating"] < valid_min) | (df["Avg_Rating"] > valid_max) | (df["Avg_Rating"].isna())]
print("Invalid Ratings:\n", invalid)

# ✅ Step 3: Correct invalid values
# Option 1: Replace with NaN
df.loc[(df["Avg_Rating"] < valid_min) | (df["Avg_Rating"] > valid_max), "Avg_Rating"] = None

# Option 2: Cap values within range
df["Avg_Rating"] = df["Avg_Rating"].clip(lower=valid_min, upper=valid_max)

print("\nCorrected DataFrame:\n", df)
```
